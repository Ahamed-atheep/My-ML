```python
def even_or_odd(num):
    if num%2==0:
        return 'The number is {} even'.format(num)
    else:
        return 'The number is {} odd'.format(num)
```

```python
lst=[1,2,3,4,5,6,7,8,9]
list(map(even_or_odd,lst))
```

```python
list(map(lambda a:a*a,lst))
```

```python
add=lambda a,b:a+b
add(3,2)
```

```python
list(filter(lambda x:x%2==0,lst))
```

```python
## list Comprehension
[i*i for i in lst if i%2==0]
```
